```python
opcion = 0
while opcion <= 6:
  print("1-. Consultar computadoras")
  print("2-. Buscar por software/Programas/Apps")
  print("3-. Solicitar prestamo")
  print("4-. Devolver computadora")
  print("5-. Reservar comptadora")
  print("6-. Agregar comptadora")
  print("7-. Salir")
  opcion = int(input("Selecciona una opcion: "))
  if opcion == 1:
    print("Seleccionaste opcion de consultar computadoras")
  elif opcion == 2:
    print("Seleccionaste la opcion de buscar por software/Programas/Apps")
  elif opcion == 3:
    print("Seleccionaste la opcion de solicitar prestamo")
  elif opcion == 4:
    print("Seleccionaste la opcion de devolver computadora")
  elif opcion == 5:
    print("Seleccionaste la opcion de reservar comptadora")
  elif opcion == 6:
    print("Seleccionaste la opcion de agregar comptadora")
  elif opcion == 7:
    print("Seleccionaste la opcion de salir")
  else:
    print("Opcion no valida")
```
