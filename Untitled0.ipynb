**Regresar un resultado - La instruccion return**

```python
def calcular_area(base, altura):
    area = base * altura
    return area
```

```python
calcular_area(25, 13)
resultado = calcular_area(25, 13)
print("Area del triangulo", resultado)
print("Area del triangulo", calcular_area(10, 5))
```

# **Alcance de las variables**
- El alcance de una varibale se refiere a la prte del codigo donde se conoce la variable
- Una variable que se define dentro de una funcion NO existe en otras funciones o fuera de la funcion
- Para extender el alcance de una variable y q pueda existir fuera de una funcion se deben definir con la palabra reservar **global**

```python
def mifuncion():
  # La variablex de mifuncion existe unicamente dentro del codigo de la funcion
  variablex = 123
  print(variablex)

mifuncion()
# La siguiente variable NO es la misma que esta definida en mifuncion, por lo tanto, no existe la variable y por eso marca error
print(variablex)
```

```python
def funcion2():
  variable1 = 5
  print("El valor de variable1 es", variable1)

variable1 =  10
print("Valor de la variable1 es", variable1)
funcion2()
print("Valor de la variable1 es", variable1)
```

```python
def funcion2():
  global variable1
  variable1 = 5
  print("El valor de variable1 es", variable1)

variable1 =  10
print("Valor de la variable1 es", variable1)
funcion2()
print("Valor de la variable1 es", variable1)
```
